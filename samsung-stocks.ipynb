```python
import pandas as pd
df=pd.read_csv("Samsung.csv")
df
```

```python
import matplotlib.pyplot as mp
small_data=df.tail(10).copy()
small_data["Moving Average"]=small_data["Close"].rolling(5).mean()
mp.plot(small_data["Date"],small_data["Close"],marker="o",label="Close Price")
mp.plot(small_data["Date"],small_data["Moving Average"],marker=".",label="Moving Average")
mp.title("Samsung stocks analysis")
mp.xlabel("Date")
mp.ylabel("Closing Prices")
mp.legend()
mp.tight_layout()
mp.grid()
mp.xticks(rotation=25)
mp.savefig("samsung_stocks.png")
mp.show()
```
